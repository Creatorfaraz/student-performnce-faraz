```python
# 6. SELECT FEATURES

features = [
    "studytime",
    "failures",
    "absences",
    "Medu",
    "Fedu",
    "freetime",
    "goout",
    "health",
    "famrel",
    "schoolsup",
    "famsup",
    "paid",
    "activities",
    "internet"
]

X = df[features]

# Regression target
y_regression = df["G3"]

# Classification target
y_classification = df["pass_fail"]



# 7. CHECK SELECTED DATA


print("\nSelected Features:")
display(X.head())

print("\nRegression Target - Final Marks:")
display(y_regression.head())

print("\nClassification Target - Pass/Fail:")
display(y_classification.head())
```

```python
# MODEL 1 - LINEAR REGRESSION
# PREDICT FINAL MARKS


# 11. TRAIN TEST SPLIT FOR REGRESSION


X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_regression,
    test_size=0.20,
    random_state=42
)

print("\nRegression Training Data:", X_train_reg.shape)
print("Regression Testing Data:", X_test_reg.shape)

```

```python
# 12. FEATURE SCALING


scaler_reg = StandardScaler()

X_train_reg_scaled = scaler_reg.fit_transform(X_train_reg)

X_test_reg_scaled = scaler_reg.transform(X_test_reg)



# 13. CREATE LINEAR REGRESSION MODEL

linear_model = LinearRegression()

linear_model.fit(
    X_train_reg_scaled,
    y_train_reg
)



# 14. MAKE PREDICTIONS


y_pred_reg = linear_model.predict(
    X_test_reg_scaled
)



# 15. EVALUATE LINEAR REGRESSION


mae = mean_absolute_error(
    y_test_reg,
    y_pred_reg
)

mse = mean_squared_error(
    y_test_reg,
    y_pred_reg
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test_reg,
    y_pred_reg
)
```

```python
print("LINEAR REGRESSION RESULTS")


print("MAE :", round(mae, 2))
print("MSE :", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 2))
```

```python
# 16. ACTUAL VS PREDICTED GRADES


comparison = pd.DataFrame({
    "Actual Grade": y_test_reg.values,
    "Predicted Grade": np.round(y_pred_reg, 2)
})

print("\nActual vs Predicted Grades:")
display(comparison.head(15))



# 17. VISUALIZE ACTUAL VS PREDICTED


plt.figure(figsize=(8, 5))

plt.scatter(
    y_test_reg,
    y_pred_reg
)

plt.xlabel("Actual Final Grade")
plt.ylabel("Predicted Final Grade")
plt.title("Actual vs Predicted Final Grades")

plt.show()
```

```python
#MODEL 2 - LOGISTIC REGRESSION
# CLASSIFY PASS / FAIL

# 18. TRAIN TEST SPLIT FOR CLASSIFICATION


X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)

print("\nClassification Training Data:", X_train_class.shape)
print("Classification Testing Data:", X_test_class.shape)
```

```python
# 19. FEATURE SCALING


scaler_class = StandardScaler()

X_train_class_scaled = scaler_class.fit_transform(
    X_train_class
)

X_test_class_scaled = scaler_class.transform(
    X_test_class
)



# 20. CREATE LOGISTIC REGRESSION MODEL


logistic_model = LogisticRegression(
    max_iter=1000
)

logistic_model.fit(
    X_train_class_scaled,
    y_train_class
)


# 21. MAKE CLASSIFICATION PREDICTIONS

y_pred_class = logistic_model.predict(
    X_test_class_scaled
)



# 22. EVALUATE CLASSIFICATION MODEL


accuracy = accuracy_score(
    y_test_class,
    y_pred_class
)

precision = precision_score(
    y_test_class,
    y_pred_class,
    zero_division=0
)

recall = recall_score(
    y_test_class,
    y_pred_class,
    zero_division=0
)
```

```python

print("LOGISTIC REGRESSION RESULTS")

print("Accuracy :", round(accuracy, 2))
print("Precision:", round(precision, 2))
print("Recall   :", round(recall, 2))

print("\nClassification Report:")
print(
    classification_report(
        y_test_class,
        y_pred_class,
        target_names=["Fail", "Pass"],
        zero_division=0
    )
)


# ------------------------------------------------------------
# 23. CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test_class,
    y_pred_class
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Fail", "Pass"],
    yticklabels=["Fail", "Pass"]
)

plt.title("Confusion Matrix - Student Pass/Fail Prediction")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

```
